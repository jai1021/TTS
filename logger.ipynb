```python
import os
import logging
from hparams import HParams as hp

class Logger(object):
    def __init__(self, dataset_name, model_name):
        self.model_name = model_name
        self.project_name = "%s-%s" % (dataset_name, self.model_name)
        self.logdir = os.path.join(hp.logdir, self.project_name)

        # Ensure the log directory exists
        os.makedirs(self.logdir, exist_ok=True)

        # Configure logging to a file
        log_file = os.path.join(self.logdir, 'training.log')
        logging.basicConfig(filename=log_file, level=logging.INFO, format='%(asctime)s - %(levelname)s: %(message)s')

    def log_step(self, phase, step, loss_dict, image_dict):
        if phase == 'train' and step % 1000 == 0:
            # Log loss values
            for key, value in loss_dict.items():
                logging.info(f'{phase} step {step} - {key}: {value}')

            # Log image information
            for key, image in image_dict.items():
                logging.info(f'{phase} step {step} - {self.model_name}/{key}: Image data')

    def log_epoch(self, phase, step, loss_dict):
        # Log loss values for each epoch
        for key, value in loss_dict.items():
            logging.info(f'{phase} epoch {step} - {key}: {value}')
```
