```python
import TTS
from TTS.utils.manage import ModelManager
from TTS.utils.synthesizer import Synthesizer
from IPython.display import Audio, display
import ipywidgets as widgets

# Your model path
path = "/Users/jaideepmanupati/anaconda3/lib/python3.11/site-packages/TTS/.models.json"

# Download and load the TTS model
model_manager = ModelManager(path)
model_path, config_path, model_item = model_manager.download_model("tts_models/en/ljspeech/tacotron2-DDC")

# Download and load the vocoder model
voc_path, voc_config_path, _ = model_manager.download_model(model_item["default_vocoder"])

# Create a synthesizer
syn = Synthesizer(
    tts_checkpoint=model_path,
    tts_config_path=config_path,
    vocoder_checkpoint=voc_path,
    vocoder_config=voc_config_path
)

# Create a text input widget
text_input = widgets.Text(value='Enter text here', description='Text:')

# Function to generate TTS and display audio
def generate_tts(button):
    user_text = text_input.value
    outputs = syn.tts(user_text)
    output_file_path = "user_audio.wav"
    syn.save_wav(outputs, output_file_path)
    display(Audio(output_file_path, autoplay=True))

# Create a button widget
generate_button = widgets.Button(description='Generate TTS')

# Attach the function to the button click event
generate_button.on_click(generate_tts)

# Display widgets
display(text_input)
display(generate_button)
```
